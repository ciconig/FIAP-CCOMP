```python
from pathlib import Path
import sys

# Localiza a pasta da atividade a partir do diretório do kernel.
for pasta in (Path.cwd(), *Path.cwd().parents):
    candidatos = (pasta, pasta / "1CCOMP" / "FASE3-ATIVIDADE")
    raiz = next(
        (candidato for candidato in candidatos
         if (candidato / "utils" / "gerar_dados.py").is_file()),
        None,
    )
    if raiz is not None:
        break
else:
    raise FileNotFoundError("Não foi possível localizar utils/gerar_dados.py")

if str(raiz) not in sys.path:
    sys.path.insert(0, str(raiz))

```

```python
from utils.gerar_dados import gerar_dados, exibir_dados

dados = gerar_dados()
historico = dados["historico"]
exibir_dados(historico)
```

```python
from utils.modelo_previsao import executar_previsao

vento_ms = 11 
resultado = executar_previsao(historico, vento_ms=vento_ms)

print("Equacao da reta:", resultado["equacao"])
print(resultado["mensagem"])
print(f"Erro medio absoluto: {resultado['ema']:.2f} kWh")
print("Proximas horas (repetindo o ultimo vento medido):")
for previsao in resultado["proximas_horas"]:
    print(
        f"Hora {previsao['hora']}: "
        f"vento {previsao['vento_ms']} m/s -> "
        f"{previsao['eolica_prevista_kwh']:.2f} kWh"
    )
```
